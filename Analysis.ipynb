```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
# Load cleaned data

df = pd.read_csv("data/cleaned_books.csv")

df.head()
```

```python
#  Descriptive statistics

print("\nPrice stats:\n", df['Price_clean'].describe())
print("\nRating stats:\n", df['Rating_clean'].describe())
```

```python
# Category frequency

cat_freq = df['Category_clean'].value_counts()
print(cat_freq.head(10))
```

```python

print("Top 5 highest prices:")
display(df.sort_values('Price_clean', ascending=False).head())

print("\nTop 5 lowest prices:")
display(df.sort_values('Price_clean', ascending=True).head())

```

```python
# --- IQR method ---
Q1_price, Q3_price = df['Price_clean'].quantile([0.25, 0.75])
IQR_price = Q3_price - Q1_price
lower_price = Q1_price - 1.5 * IQR_price
upper_price = Q3_price + 1.5 * IQR_price
price_outliers = df[(df['Price_clean'] < lower_price) | (df['Price_clean'] > upper_price)]

Q1_rating, Q3_rating = df['Rating_clean'].quantile([0.25, 0.75])
IQR_rating = Q3_rating - Q1_rating
lower_rating = Q1_rating - 1.5 * IQR_rating
upper_rating = Q3_rating + 1.5 * IQR_rating
rating_outliers = df[(df['Rating_clean'] < lower_rating) | (df['Rating_clean'] > upper_rating)]

print("Price outliers (IQR):", len(price_outliers))
print("Rating outliers (IQR):", len(rating_outliers))

# --- Z-score method ---
df['Price_z'] = stats.zscore(df['Price_clean'])
df['Rating_z'] = stats.zscore(df['Rating_clean'])

price_outliers_z = df[np.abs(df['Price_z']) > 3]
rating_outliers_z = df[np.abs(df['Rating_z']) > 3]

print("Price outliers (Z-score):", len(price_outliers_z))
print("Rating outliers (Z-score):", len(rating_outliers_z))
```

```python
# Correlation
df['Availability_clean'] = df['Availability'].str.contains("In stock", case=False).astype(int)
corr = df[['Price_clean','Rating_clean','Availability_clean']].corr()
print(corr)
```

```python
#Hypothesis

# Inspect all categories
print("All categories in dataset:")
display(df['Category_clean'].value_counts())

# Choose categories that actually exist
cat1 = 'fiction'
cat2 = 'historical fiction'  

# Select numeric price values
group1 = df[df['Category_clean'] == cat1]['Price_clean']
group2 = df[df['Category_clean'] == cat2]['Price_clean']

print(f"\nGroup1 ({cat1}) size: {len(group1)}")
print(f"Group2 ({cat2}) size: {len(group2)}")

# Run T-test only if both groups have enough data
if len(group1) >= 2 and len(group2) >= 2:
    t_stat, p_val = stats.ttest_ind(group1, group2, nan_policy='omit')
    print(f"T-test between '{cat1}' and '{cat2}': t={t_stat:.2f}, p={p_val:.4f}")
```
