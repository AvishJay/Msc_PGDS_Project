```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import plotly.express as px

df = pd.read_csv("data/cleaned_books.csv")
```

```python
# Price histogram by category (top 5)

top_cat = df['Category_clean'].value_counts().head(5).index
sns.histplot(data=df[df['Category_clean'].isin(top_cat)], x='Price_clean', hue='Category_clean', multiple='stack')
plt.show()
```

```python
# Boxplot Price by Category

sns.boxplot(data=df[df['Category_clean'].isin(top_cat)], x='Category_clean', y='Price_clean')
plt.show()
```

```python
# Scatter Rating vs Price

sns.scatterplot(data=df, x='Price_clean', y='Rating_clean', hue='Category_clean')
plt.show()
```

```python
# Interactive Plotly bar

fig = px.bar(df.groupby('Category_clean')['Price_clean'].mean().reset_index(),
             x='Category_clean', y='Price_clean', title='Average Price by Category')
fig.show()
```

```python
#Bar chart: Category popularity

category_counts = df['Category_clean'].value_counts()

plt.figure(figsize=(12,6))
sns.barplot(x=category_counts.index, y=category_counts.values, palette="viridis")
plt.xticks(rotation=45, ha='right')
plt.title("Category Popularity (Number of Books)")
plt.ylabel("Number of Books")
plt.xlabel("Category")
plt.show()
```

```python
#Bar chart: Average ratings by category

avg_rating = df.groupby('Category_clean')['Rating_clean'].mean().sort_values(ascending=False)

plt.figure(figsize=(12,6))
sns.barplot(x=avg_rating.index, y=avg_rating.values, palette="coolwarm")
plt.xticks(rotation=45, ha='right')
plt.title("Average Rating by Category")
plt.ylabel("Average Rating")
plt.xlabel("Category")
plt.show()
```
